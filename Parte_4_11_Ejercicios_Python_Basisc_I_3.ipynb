## Ejercicio 1
Tenemos las siguientes variables:

```Python

B = "Text"

```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
A = 4
B = "Text"
C = 4.1

print(A == B)
print(A == C)
print(A > C)
print(C <= A)
print(B == C)
````

## Ejercicio 2
Crea un programa donde se recojan dos inputs del usuario, y el output del programa sea si esos inputs son iguales o no

````python
nombre_usuario = input("¿Cómo te llamas? ")
nombre_madre = input("¿Cómo se llama tu madre ")

if(nombre_usuario == nombre_madre):
    print(nombre_usuario,", tu nombre coincide con el de tu madre")
else:
    print(nombre_usuario,", tu nombre no coincide con el de tu madre", nombre_madre)
````
