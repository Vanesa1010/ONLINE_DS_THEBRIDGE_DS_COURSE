## Ejercicio 2
Haz un programa que detecte todos los duplicados de un elemento en una lista:
* Utiliza una variable duplicado para asignarle el valor del elemento del que queremos encontrar todos sus duplicados.
* El programa debe devolver todos los índices de los elementos duplicados.
* Aplícalo para encontrar los duplicados de "un", "es" y "binario" de:  

>```ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]```


`````python
def busca_duplicados(lista, duplicado):
    indices = []
    for i in range(len(lista)):
        if (lista[i] == duplicado):
            indices.append(i)
    return indices

ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]
print("La palabra 'un' está duplicada en la posición", busca_duplicados(ejer_3, "un"))
print("La palabra 'es' está duplicada en la posición", busca_duplicados(ejer_3, "es"))
print("La palabra 'binario' está duplicada en la posición", busca_duplicados(ejer_3, "binario"))
`````

`````python
def busca_duplicados(lista):
    indices = []
    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            if (lista[i] == lista[j]):
                indices.append(i)
                indices.append(j)
    print("los elementos duplicados están en las posiciones", indices)
    return indices

ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]
print(busca_duplicados(ejer_3))
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


`````python
def dni_valido(dni, letra):
    palabra='TRWAGMYFPDXBNJZSQVHLCKE'
    if dni.isdigit() and len(dni) >= 7 and len(dni) <= 8:
        if palabra[int(dni) % 23] == letra.upper():
            print("Dni valido", dni, letra)
    else:
        print("Dni no válido")

dni = input("Introduce el número de tu DNI: ")
letra = input("Introduce la letra: ")
dni_valido(dni, letra)
`````
