## Ejercicio 1
Para este ejercicio vamos a poner en práctica [las funciones built in](https://docs.python.org/3/library/functions.html).

1. Calcula el máximo de la lista: [4, 6, 8, -1]
2. Suma todos los elementos de la lista anterior
3. Redondea este float a 3 dígitos decimales: 63.451256965
4. Valor absoluto de: -74

```python
# Máximo
lista = [4, 6, 8, -1]
máximo = max(lista)
print(máximo)

# Suma todos los elementos de la lista 
suma = lista[0] + lista[1] + lista[2] + lista[3]
print(suma)

# Redondea
num_decimal = 63.451256965
resultado_redondeado = round(num_decimal, 3)
print(resultado_redondeado)

# Valor absoluto
numero = -74
print(abs(numero)) 
```

## Ejercicio 2
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

```python
# Mayúsculas
cadena = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
mayusculas = cadena.upper()
print(mayusculas)

# Minúsculas
minusculas = cadena.lower()
print(minusculas)

# Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
prim_mayus = cadena.title()
print(prim_mayus)

# lista dividida por sus espacios
lista_espacios = cadena.split(" ")
print(lista_espacios)

# Sustituy las comas , por puntos y comas ;
nueva_cadena = cadena.replace(",",";")
print(nueva_cadena)

# Elimina las a minúsculas
nueva_cadena_2 = cadena.replace("a","")
print(nueva_cadena_2)
nueva_cadena_3 = prim_mayus.replace("a","")
print(nueva_cadena_3)
```

## Ejercicio 3
1. Crea una lista con 3 elementos numéricos
2. Añade un cuarto elemento
3. Calcula la suma de todos
4. Elimina el segundo elemento de la lista
5. Añade otro elemento en la posicion 3 de la lista
6. Crea otra lista con 4 elementos y concaténala a la que ya tenías.
7. Ordena la lista de menor a mayor.
8. BONUS: Ordena la lista de mayor a menor.  
(*PISTA: Para hacer 8, tendrás que buscar en la información del método...*)  
  
  
NOTA: Pinta por pantalla el resultado de cada paso.

```python
lista_nums = [7, 27, 55.25]
print(lista_nums)

# Añadir un cuarto elemento
nueva_lista = lista_nums.append(55)
print(lista_nums)

# Suma elementos lista
suma = 0
for indice in lista_nums:
    suma +=indice
print("La suma de los 4 elementos de la lista es", suma)

# Elimina el segundo elemento de la lista
del lista_nums[1]
print(lista_nums)

# Añade otro elemento en la posicion 3 de la lista
lista_nums.insert(3, 88.271)
print(lista_nums)

# Crea otra lista con 4 elementos y concaténala a la que ya tenías
lista_2 = [-17, 255, 212, 3333]
lista_ampliada = lista_nums + lista_2
print(lista_ampliada)

# Ordena la lista de menor a mayor
lista_creciente = sorted(lista_nums)
print(lista_creciente)

# Ordena la lista de mayor a menor
lista_decreciente = sorted(lista_nums, reverse=True)
print(lista_decreciente)
```
